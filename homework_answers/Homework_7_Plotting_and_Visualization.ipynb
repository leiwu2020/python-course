## Exercise 1: Basic Plotting with Matplotlib (20 points)

**Task:** Create basic plots using Matplotlib.

**Requirements:**
- Create line plots, bar charts, and scatter plots
- Customize plot appearance
- Add labels, titles, and legends
- Use different plot styles


```python
# TODO: Complete the following tasks

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

# Set style for better-looking plots
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# Load sample data
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df['species'] = iris.target_names[iris.target]

# 1. Create line plots, bar charts, and scatter plots
print("=== BASIC PLOTTING ===")

# Line plot
plt.figure(figsize=(10, 6))
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label='sin(x)', linewidth=2)
plt.plot(x, y2, label='cos(x)', linewidth=2)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Trigonometric Functions')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Bar chart
plt.figure(figsize=(10, 6))
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]

bars = plt.bar(categories, values, color=['red', 'green', 'blue', 'orange', 'purple'])
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Sample Bar Chart')
plt.grid(True, alpha=0.3)

# Add value labels on bars
for bar, value in zip(bars, values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1, 
             str(value), ha='center', va='bottom')

plt.show()

# Scatter plot
plt.figure(figsize=(10, 6))
x_scatter = np.random.randn(100)
y_scatter = 2 * x_scatter + np.random.randn(100)

plt.scatter(x_scatter, y_scatter, alpha=0.6, c='blue', s=50)
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Sample Scatter Plot')
plt.grid(True, alpha=0.3)
plt.show()

# 2. Customize plot appearance
print("\n=== PLOT CUSTOMIZATION ===")

# TODO: Create a customized plot
plt.figure(figsize=(12, 8))

# Create subplots
plt.subplot(2, 2, 1)
plt.plot(x, y1, 'r-', linewidth=2, label='sin(x)')
plt.title('Subplot 1: Sine Wave')
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 2)
plt.bar(categories, values, color='skyblue', edgecolor='navy', linewidth=2)
plt.title('Subplot 2: Bar Chart')
plt.xticks(rotation=45)

plt.subplot(2, 2, 3)
plt.scatter(x_scatter, y_scatter, alpha=0.7, c='green', s=60)
plt.title('Subplot 3: Scatter Plot')
plt.grid(True)

plt.subplot(2, 2, 4)
# TODO: Create a histogram
data_hist = np.random.normal(0, 1, 1000)
plt.hist(data_hist, bins=30, alpha=0.7, color='orange', edgecolor='black')
plt.title('Subplot 4: Histogram')
plt.grid(True)

plt.tight_layout()
plt.show()

# 3. Use Seaborn for statistical visualization
print("\n=== SEABORN STATISTICAL PLOTS ===")

# Pair plot
plt.figure(figsize=(12, 10))
sns.pairplot(iris_df, hue='species', diag_kind='hist')
plt.suptitle('Iris Dataset Pair Plot', y=1.02)
plt.show()

# Heatmap
plt.figure(figsize=(10, 8))
correlation_matrix = iris_df.select_dtypes(include=[np.number]).corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.show()

# Box plot
plt.figure(figsize=(12, 6))
iris_melted = iris_df.melt(id_vars=['species'], 
                          value_vars=iris.feature_names,
                          var_name='feature', value_name='value')
sns.boxplot(data=iris_melted, x='feature', y='value', hue='species')
plt.title('Box Plot of Iris Features by Species')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 4. Create publication-quality graphics
print("\n=== PUBLICATION-QUALITY GRAPHICS ===")

# TODO: Create a professional-looking plot
fig, ax = plt.subplots(figsize=(12, 8))

# Create multiple datasets
x = np.linspace(0, 4*np.pi, 100)
y1 = np.sin(x)
y2 = np.sin(x + np.pi/4)
y3 = np.sin(x + np.pi/2)

# Plot with custom styling
ax.plot(x, y1, 'o-', label='sin(x)', linewidth=2, markersize=4, alpha=0.8)
ax.plot(x, y2, 's-', label='sin(x + π/4)', linewidth=2, markersize=4, alpha=0.8)
ax.plot(x, y3, '^-', label='sin(x + π/2)', linewidth=2, markersize=4, alpha=0.8)

# Customize appearance
ax.set_xlabel('x (radians)', fontsize=12, fontweight='bold')
ax.set_ylabel('y', fontsize=12, fontweight='bold')
ax.set_title('Phase-Shifted Sine Waves', fontsize=14, fontweight='bold')
ax.legend(fontsize=11, framealpha=0.9)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_xlim(0, 4*np.pi)

# Add annotations
ax.annotate('Peak', xy=(np.pi/2, 1), xytext=(np.pi/2 + 0.5, 1.2),
            arrowprops=dict(arrowstyle='->', color='red', lw=2),
            fontsize=10, color='red', fontweight='bold')

plt.tight_layout()
plt.show()

print("All plotting exercises completed!")
```
