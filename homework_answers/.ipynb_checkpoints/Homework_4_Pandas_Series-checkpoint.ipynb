## Exercise 1: Series Creation and Basic Operations (20 points)

**Task:** Create Pandas Series and perform basic operations.

**Requirements:**
- Create Series from different data sources
- Perform basic operations on Series
- Use Series indexing and selection
- Demonstrate Series methods


```python
# TODO: Complete the following tasks

import pandas as pd
import numpy as np

# 1. Create Series from different data sources
print("=== SERIES CREATION ===")

# Create Series from list
data_list = [10, 20, 30, 40, 50]
series1 = pd.Series(data_list)
print(f"Series from list:\n{series1}")

# TODO: Create Series with custom index
custom_index = ['A', 'B', 'C', 'D', 'E']
series2 = pd.Series(data_list, index=custom_index)
print(f"\nSeries with custom index:\n{series2}")

# Create Series from dictionary
data_dict = {'Alice': 25, 'Bob': 30, 'Charlie': 35, 'Diana': 28}
series3 = pd.Series(data_dict)
print(f"\nSeries from dictionary:\n{series3}")

# Create Series from NumPy array
np_array = np.array([1.5, 2.5, 3.5, 4.5, 5.5])
series4 = pd.Series(np_array, name='Values')
print(f"\nSeries from NumPy array:\n{series4}")

# 2. Perform basic operations on Series
print(f"\n=== BASIC OPERATIONS ===")

# Arithmetic operations
print(f"Series1 + 10:\n{series1 + 10}")
print(f"Series1 * 2:\n{series1 * 2}")
print(f"Series1 + Series2:\n{series1 + series2}")

# Statistical operations
print(f"\nStatistical operations:")
print(f"Mean: {series1.mean()}")
print(f"Sum: {series1.sum()}")
print(f"Max: {series1.max()}")
print(f"Min: {series1.min()}")
print(f"Std: {series1.std():.2f}")

# 3. Use Series indexing and selection
print(f"\n=== INDEXING AND SELECTION ===")

# Position-based indexing
print(f"First element: {series1[0]}")
print(f"Last element: {series1[-1]}")
print(f"First 3 elements: {series1[:3]}")

# Label-based indexing
print(f"\nLabel-based indexing:")
print(f"Element 'B': {series2['B']}")
print(f"Elements 'A' to 'C': {series2['A':'C']}")

# Boolean indexing
print(f"\nBoolean indexing:")
mask = series1 > 25
print(f"Elements > 25: {series1[mask]}")

# 4. Demonstrate Series methods
print(f"\n=== SERIES METHODS ===")

# String methods (if applicable)
text_series = pd.Series(['apple', 'banana', 'cherry', 'date'])
print(f"Original: {text_series}")
print(f"Uppercase: {text_series.str.upper()}")
print(f"Length: {text_series.str.len()}")

# Mathematical methods
print(f"\nMathematical methods:")
print(f"Absolute values: {pd.Series([-1, -2, 3, -4]).abs()}")
print(f"Square root: {pd.Series([4, 9, 16, 25]).apply(np.sqrt)}")

# Sorting
print(f"\nSorting:")
print(f"Sorted by values: {series1.sort_values()}")
print(f"Sorted by index: {series2.sort_index()}")
```
