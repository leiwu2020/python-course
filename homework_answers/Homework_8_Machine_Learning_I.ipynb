## Exercise 1: Data Preprocessing and Exploration (20 points)

**Task:** Preprocess data for machine learning and perform exploratory data analysis.

**Requirements:**
- Load and explore datasets
- Handle missing values and outliers
- Perform feature scaling and encoding
- Split data into training and testing sets


```python
# TODO: Complete the following tasks

import pandas as pd
import numpy as np
from sklearn.datasets import load_iris, load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load and explore datasets
print("=== DATASET LOADING AND EXPLORATION ===")

# Load Iris dataset
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target
iris_df['species'] = iris.target_names[iris.target]

print(f"Iris dataset shape: {iris_df.shape}")
print(f"\nFirst 5 rows:\n{iris_df.head()}")
print(f"\nDataset info:")
print(iris_df.info())
print(f"\nStatistical summary:\n{iris_df.describe()}")

# TODO: Load Wine dataset
wine = load_wine()
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)
wine_df['target'] = wine.target
wine_df['wine_type'] = wine.target_names[wine.target]

print(f"\nWine dataset shape: {wine_df.shape}")
print(f"\nFirst 5 rows:\n{wine_df.head()}")

# 2. Handle missing values and outliers
print(f"\n=== DATA CLEANING ===")

# Check for missing values
print(f"Missing values in Iris dataset:\n{iris_df.isnull().sum()}")
print(f"Missing values in Wine dataset:\n{wine_df.isnull().sum()}")

# TODO: Handle outliers using IQR method
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers

# Check for outliers in numerical columns
numerical_columns = iris_df.select_dtypes(include=[np.number]).columns
print(f"\nOutlier detection for Iris dataset:")
for col in numerical_columns:
    if col != 'target':
        outliers = detect_outliers_iqr(iris_df, col)
        print(f"{col}: {len(outliers)} outliers")

# 3. Perform feature scaling and encoding
print(f"\n=== FEATURE SCALING AND ENCODING ===")

# Separate features and target
X_iris = iris_df.drop(['target', 'species'], axis=1)
y_iris = iris_df['target']

# TODO: Apply StandardScaler
scaler = StandardScaler()
X_iris_scaled = scaler.fit_transform(X_iris)
X_iris_scaled_df = pd.DataFrame(X_iris_scaled, columns=X_iris.columns)

print(f"Original features (first 5 rows):\n{X_iris.head()}")
print(f"Scaled features (first 5 rows):\n{X_iris_scaled_df.head()}")

# 4. Split data into training and testing sets
print(f"\n=== DATA SPLITTING ===")

# TODO: Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X_iris_scaled, y_iris, test_size=0.2, random_state=42, stratify=y_iris
)

print(f"Training set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
print(f"Training target distribution: {np.bincount(y_train)}")
print(f"Test target distribution: {np.bincount(y_test)}")

# 5. Train and evaluate models
print(f"\n=== MODEL TRAINING AND EVALUATION ===")

# Logistic Regression
lr_model = LogisticRegression(random_state=42)
lr_model.fit(X_train, y_train)
lr_pred = lr_model.predict(X_test)
lr_accuracy = accuracy_score(y_test, lr_pred)

print(f"Logistic Regression Accuracy: {lr_accuracy:.4f}")

# TODO: Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_pred)

print(f"Random Forest Accuracy: {rf_accuracy:.4f}")

# Model comparison
print(f"\nModel Comparison:")
print(f"Logistic Regression: {lr_accuracy:.4f}")
print(f"Random Forest: {rf_accuracy:.4f}")

# Classification report
print(f"\nRandom Forest Classification Report:")
print(classification_report(y_test, rf_pred, target_names=iris.target_names))
```
