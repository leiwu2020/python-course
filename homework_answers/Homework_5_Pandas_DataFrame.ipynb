## Exercise 1: DataFrame Creation and Basic Operations (20 points)

**Task:** Create Pandas DataFrames and perform basic operations.

**Requirements:**
- Create DataFrames from different data sources
- Perform basic operations on DataFrames
- Use DataFrame indexing and selection
- Demonstrate DataFrame methods


```python
# TODO: Complete the following tasks

import pandas as pd
import numpy as np

# 1. Create DataFrames from different data sources
print("=== DATAFRAME CREATION ===")

# Create DataFrame from dictionary
data_dict = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [25, 30, 35, 28, 32],
    'City': ['New York', 'London', 'Tokyo', 'Paris', 'Sydney'],
    'Salary': [50000, 60000, 70000, 55000, 65000]
}
df1 = pd.DataFrame(data_dict)
print(f"DataFrame from dictionary:\n{df1}")

# TODO: Create DataFrame from list of lists
data_list = [
    ['Alice', 25, 'New York', 50000],
    ['Bob', 30, 'London', 60000],
    ['Charlie', 35, 'Tokyo', 70000],
    ['Diana', 28, 'Paris', 55000],
    ['Eve', 32, 'Sydney', 65000]
]
columns = ['Name', 'Age', 'City', 'Salary']
df2 = pd.DataFrame(data_list, columns=columns)
print(f"\nDataFrame from list of lists:\n{df2}")

# Create DataFrame from NumPy array
np_data = np.random.randint(1, 100, (5, 4))
df3 = pd.DataFrame(np_data, columns=['A', 'B', 'C', 'D'])
print(f"\nDataFrame from NumPy array:\n{df3}")

# 2. Perform basic operations on DataFrames
print(f"\n=== BASIC OPERATIONS ===")

# DataFrame info
print(f"DataFrame shape: {df1.shape}")
print(f"DataFrame columns: {list(df1.columns)}")
print(f"DataFrame index: {list(df1.index)}")
print(f"DataFrame dtypes:\n{df1.dtypes}")

# Statistical summary
print(f"\nStatistical summary:")
print(df1.describe())

# 3. Use DataFrame indexing and selection
print(f"\n=== INDEXING AND SELECTION ===")

# Column selection
print(f"Name column: {df1['Name']}")
print(f"Age and Salary columns:\n{df1[['Age', 'Salary']]}")

# Row selection
print(f"\nFirst row:\n{df1.iloc[0]}")
print(f"First 3 rows:\n{df1.iloc[:3]}")

# Label-based selection
print(f"\nLabel-based selection:")
print(f"Row 0:\n{df1.loc[0]}")
print(f"Rows 0-2, columns 'Name' and 'Age':\n{df1.loc[0:2, ['Name', 'Age']]}")

# Boolean indexing
print(f"\nBoolean indexing:")
high_salary = df1['Salary'] > 60000
print(f"High salary employees:\n{df1[high_salary]}")

# 4. Demonstrate DataFrame methods
print(f"\n=== DATAFRAME METHODS ===")

# Data cleaning
print(f"Missing values:\n{df1.isnull().sum()}")

# Sorting
print(f"\nSorted by Age:\n{df1.sort_values('Age')}")
print(f"Sorted by Salary (descending):\n{df1.sort_values('Salary', ascending=False)}")

# Grouping
print(f"\nGroup by City:\n{df1.groupby('City')['Salary'].mean()}")

# Adding new columns
df1['Age_Group'] = df1['Age'].apply(lambda x: 'Young' if x < 30 else 'Old')
print(f"\nDataFrame with Age_Group:\n{df1}")
```
