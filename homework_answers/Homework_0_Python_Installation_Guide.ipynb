## Exercise 1: Python Version Verification (10 points)

**Task:** Check your Python installation and display version information.

**Requirements:**
- Display Python version
- Display Python executable path
- Verify Python version is 3.8 or higher
- Display platform information


```python
# Your code here
import sys
import platform

# TODO: Complete the following tasks:
# 1. Print Python version
# 2. Print Python executable path
# 3. Check if Python version is 3.8 or higher
# 4. Print platform information

print("Python Version:", sys.version)
print("\nPython Executable:", sys.executable)
print("\nPlatform:", platform.platform())
print("\nArchitecture:", platform.architecture())

# Version check
version_info = sys.version_info
if version_info.major >= 3 and version_info.minor >= 8:
    print("\n✅ Python version is compatible (3.8+)")
else:
    print("\n❌ Python version is too old. Please upgrade to 3.8 or higher.")
```

## Exercise 2: Virtual Environment Setup (15 points)

**Task:** Create and activate a virtual environment for this course.

**Requirements:**
- Create a virtual environment named 'python-course-env'
- Activate the virtual environment
- Verify the environment is active
- Display the Python path in the virtual environment


```python
# Note: Virtual environment creation should be done in terminal/command prompt
# Run these commands in your terminal:

print("To create a virtual environment, run these commands in your terminal:")
print("\n1. Create virtual environment:")
print("   python -m venv python-course-env")

print("\n2. Activate virtual environment:")
print("   # On Windows:")
print("   python-course-env\\Scripts\\activate")
print("   # On macOS/Linux:")
print("   source python-course-env/bin/activate")

print("\n3. Verify activation:")
print("   which python  # macOS/Linux")
print("   where python  # Windows")

# After activation, run this cell to verify
import sys
print("\nCurrent Python path:", sys.executable)
print("Virtual environment active:", 'python-course-env' in sys.executable)
```
