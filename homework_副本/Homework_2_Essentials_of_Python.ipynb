**Task:** Work with Python's built-in data structures and demonstrate their operations.

**Requirements:**
- Create and manipulate lists, tuples, dictionaries, and sets

- Use list comprehensions and generator expressions

```python
# TODO: Complete the following tasks

# 1. Lists - Dynamic arrays
print("=== LISTS ===")

# Create lists
numbers = [1, 2, 3, 4, 5]
fruits = ['apple', 'banana', 'cherry', 'date']
mixed = [1, 'hello', 3.14, True, [1, 2, 3]]

print(f"Numbers list: {numbers}")
print(f"Fruits list: {fruits}")
print(f"Mixed list: {mixed}")

# TODO: List operations
print(f"\nList Operations:")
print(f"Length: {_______}") # get length of the number list
print(f"First element: {numbers[0]}") # get the first number
print(f"Last element: {numbers[-1]}")
print(f"Slice [1:3]: {numbers[1:3]}")

# List methods
numbers.append(6)
print(f"After append(6): {numbers}")

numbers.insert(2, 2.5)
print(f"After insert(2, 2.5): {numbers}")

numbers.remove(2.5)
print(f"After remove(2.5): {numbers}")

popped = numbers.pop()
print(f"Popped element: {popped}, List: {numbers}")

# List comprehensions
squares = [x**2 for x in range(1, 6)]
print(f"Squares: {squares}")

even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(f"Even squares: {even_squares}")

# 2. Tuples - Immutable sequences
print(f"\n=== TUPLES ===")

# Create tuples
coordinates = (3, 4)
person = ('Alice', 25, 'Engineer')
colors = ('red', 'green', 'blue')

print(f"Coordinates: {coordinates}")
print(f"Person: {person}")
print(f"Colors: {colors}")

# TODO: Tuple operations
print(f"\nTuple Operations:")
print(f"Length: {len(coordinates)}")
print(f"First element: {coordinates[0]}")
print(f"Last element: {coordinates[-1]}")

# Tuple unpacking
name, age, job = person
print(f"Unpacked - Name: {name}, Age: {age}, Job: {job}")

# Tuple methods
print(f"Count of 'red' in colors: {colors.count('red')}")
print(f"Index of 'green' in colors: {colors.index('green')}")

# 3. Dictionaries - Key-value pairs
print(f"\n=== DICTIONARIES ===")

# Create dictionaries
student = {
    'name': 'Bob',
    'age': 22,
    'major': 'Computer Science',
    'grades': [85, 92, 78, 96]
}

scores = dict(math=95, science=87, english=92)

print(f"Student: {student}")
print(f"Scores: {scores}")

# TODO: Dictionary operations
print(f"\nDictionary Operations:")
print(f"Student name: {student['name']}")
print(f"Student age: {student.get('age', 'Not specified')}")

# Add new key-value pair
student['year'] = 'Senior'
print(f"After adding year: {student}")

# Update existing value
student['age'] = 23
print(f"After updating age: {student}")

# Dictionary methods
print(f"Keys: {list(student.keys())}")
print(f"Values: {list(student.values())}")
print(f"Items: {list(student.items())}")

# Dictionary comprehension
squared_scores = {subject: score**2 for subject, score in scores.items()}
print(f"Squared scores: {squared_scores}")

# 4. Sets - Unique elements
print(f"\n=== SETS ===")

# Create sets
numbers_set = {1, 2, 3, 4, 5}
fruits_set = {'apple', 'banana', 'cherry', 'date'}
mixed_set = {1, 'hello', 3.14, True}

print(f"Numbers set: {numbers_set}")
print(f"Fruits set: {fruits_set}")
print(f"Mixed set: {mixed_set}")

# TODO: Set operations
print(f"\nSet Operations:")
print(f"Length: {len(numbers_set)}")
print(f"Is 3 in set: {3 in numbers_set}")

# Set methods
numbers_set.add(6)
print(f"After add(6): {numbers_set}")

numbers_set.remove(6)
print(f"After remove(6): {numbers_set}")

# Set operations
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print(f"Set1: {set1}")
print(f"Set2: {set2}")
print(f"Union: {set1 | set2}")
print(f"Intersection: {set1 & set2}")
print(f"Difference: {set1 - set2}")
print(f"Symmetric difference: {set1 ^ set2}")

# 5. Data structure transformations
print(f"\n=== DATA STRUCTURE TRANSFORMATIONS ===")

# List to set (removes duplicates)
numbers_with_duplicates = [1, 2, 2, 3, 3, 3, 4, 5]
unique_numbers = list(set(numbers_with_duplicates))
print(f"Original: {numbers_with_duplicates}")
print(f"Unique: {unique_numbers}")

# Dictionary to list of tuples
student_items = list(student.items())
print(f"Student as list of tuples: {student_items}")

# List of tuples to dictionary
pairs = [('a', 1), ('b', 2), ('c', 3)]
pairs_dict = dict(pairs)
print(f"Pairs as dictionary: {pairs_dict}")

# TODO: Create a nested data structure
nested_data = {
    'students': [
        {'name': 'Alice', 'grades': [85, 92, 78]},
        {'name': 'Bob', 'grades': [90, 88, 95]},
        {'name': 'Charlie', 'grades': [75, 82, 88]}
    ],
    'course': 'Python Programming',
    'instructor': 'Dr. Smith'
}

print(f"\nNested data structure:")
print(f"Course: {nested_data['course']}")
print(f"Instructor: {nested_data['instructor']}")
print(f"Students: {nested_data['students']}")

# Access nested data
first_student = nested_data['students'][0]
print(f"First student: {first_student}")
print(f"First student's name: {first_student['name']}")
print(f"First student's grades: {first_student['grades']}")
```
