## 2. Basic Data Types


### Understanding Python Data Types

Python is a dynamically typed language, which means you don't need to explicitly declare the type of a variable. Python automatically determines the data type based on the value assigned. Let's explore the fundamental data types in Python:

**Key Data Types:**
- **Integer (int)**: Whole numbers (positive, negative, or zero)
- **Float**: Decimal numbers with fractional parts
- **String (str)**: Text data enclosed in quotes
- **Boolean (bool)**: True or False values
- **None**: Represents the absence of a value

**Why This Matters:**
Understanding data types is crucial because:
- Different types have different operations available
- Type conversion is often necessary when working with data
- Knowing the type helps debug errors and write better code

The code below demonstrates how to create variables of different types and use the `type()` function to check their data types.


````python
# Integer
age = 25
print(f"Age: {age}, Type: {type(age)}")

# Float
height = 5.9
print(f"Height: {height}, Type: {type(height)}")

# String
name = "Alice"
print(f"Name: {name}, Type: {type(name)}")

# Boolean
is_student = True
print(f"Is student: {is_student}, Type: {type(is_student)}")

# None
data = None
print(f"Data: {data}, Type: {type(data)}")
````

## 3. Variables and Assignment


### Working with Variables and Assignment

Variables are containers that store data in memory. In Python, variable assignment is straightforward and flexible. Here's what we'll learn:

**Key Concepts:**
- **Simple Assignment**: `variable = value`
- **Multiple Assignment**: Assigning multiple values to multiple variables in one line
- **Variable Swapping**: Python's elegant way to swap two variables without a temporary variable
- **Augmented Assignment**: Shortcut operators like `+=`, `-=`, `*=`, `/=`
- **String Formatting**: Using f-strings for dynamic string creation

**Important Notes:**
- Variable names should be descriptive and follow Python naming conventions
- Python variable names are case-sensitive
- Variables can be reassigned to different data types (dynamic typing)
- Use meaningful names that describe what the variable represents

The code below demonstrates various assignment techniques and string formatting methods that you'll use frequently in Python programming.


````python
# Variable assignment
x = 10
y = 20

# Multiple assignment
a, b, c = 1, 2, 3
print(f"a={a}, b={b}, c={c}")

# Swapping variables
x, y = y, x
print(f"After swap: x={x}, y={y}")

# Augmented assignment
count = 5
count += 3  # Same as count = count + 3
print(f"Count after += 3: {count}")

# String formatting
first_name = "John"
last_name = "Doe"
full_name = f"{first_name} {last_name}"
print(f"Full name: {full_name}")
````

## 4. Python Operators: Arithmetic, Comparison, and Logical

Operators are special symbols that perform operations on variables and values. Python provides several categories of operators:

**Arithmetic Operators:**
- `+` Addition
- `-` Subtraction  
- `*` Multiplication
- `/` Division (returns float)
- `//` Floor division (returns integer)
- `%` Modulus (remainder)
- `**` Exponentiation (power)

**Comparison Operators:**
- `==` Equal to
- `!=` Not equal to
- `<` Less than
- `>` Greater than
- `<=` Less than or equal to
- `>=` Greater than or equal to

**Logical Operators:**
- `and` Logical AND (both conditions must be True)
- `or` Logical OR (at least one condition must be True)
- `not` Logical NOT (inverts the boolean value)

**Key Points to Remember:**
- Comparison operators return boolean values (True/False)
- Logical operators work with boolean values and support short-circuit evaluation
- Division in Python always returns a float, use floor division (`//`) for integer results
- Operator precedence matters - use parentheses to clarify complex expressions

The code below demonstrates all these operators with practical examples.


````python
# Arithmetic operators
a, b = 10, 3

print(f"Addition: {a} + {b} = {a + b}")
print(f"Subtraction: {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division: {a} / {b} = {a / b}")
print(f"Floor division: {a} // {b} = {a // b}")
print(f"Modulus: {a} % {b} = {a % b}")
print(f"Exponentiation: {a} ** {b} = {a ** b}")

# Comparison operators
x, y = 5, 8
print(f"\nComparison operators:")
print(f"{x} == {y}: {x == y}")
print(f"{x} != {y}: {x != y}")
print(f"{x} < {y}: {x < y}")
print(f"{x} > {y}: {x > y}")
print(f"{x} <= {y}: {x <= y}")
print(f"{x} >= {y}: {x >= y}")

# Logical operators
p, q = True, False
print(f"\nLogical operators:")
print(f"p and q: {p and q}")
print(f"p or q: {p or q}")
print(f"not p: {not p}")
````

## 5. Conditional Statements: Making Decisions in Python

Conditional statements allow your program to make decisions based on different conditions. They are fundamental to creating dynamic and responsive programs.

**Types of Conditional Statements:**

**1. if Statement:**
- Executes code only if a condition is True
- Basic structure: `if condition:`

**2. if-else Statement:**
- Executes one block if condition is True, another if False
- Structure: `if condition: ... else: ...`

**3. if-elif-else Statement:**
- Handles multiple conditions in sequence
- Structure: `if condition1: ... elif condition2: ... else: ...`
- Only the first True condition executes

**4. Ternary Operator (Conditional Expression):**
- Compact way to write simple if-else statements
- Format: `value_if_true if condition else value_if_false`

**5. Nested Conditions:**
- Conditions inside other conditions
- Allows for complex decision-making logic

**Important Syntax Rules:**
- Use `:` after the condition
- Indent code blocks consistently (4 spaces recommended)
- Python uses indentation to define code blocks (no braces like other languages)
- Conditions are evaluated in order until one is True

The code below demonstrates all these conditional statement types with practical examples including grade calculation and weather-based decisions.


````python
# if-else statements
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")

# Ternary operator
age = 20
status = "adult" if age >= 18 else "minor"
print(f"Age: {age}, Status: {status}")

# Nested conditions
temperature = 25
is_sunny = True

if temperature > 20:
    if is_sunny:
        print("Great weather for outdoor activities!")
    else:
        print("Warm but cloudy day")
else:
    print("It's a bit cold today")
````

## 6. Loops: Repeating Code Efficiently

Loops are essential for executing code repeatedly, which is one of the most powerful features in programming. Python provides two main types of loops:

**1. for Loop:**
- Used when you know how many times you want to repeat code
- Iterates over sequences (lists, strings, ranges, etc.)
- Structure: `for variable in sequence:`

**2. while Loop:**
- Used when you want to repeat code until a condition becomes False
- Structure: `while condition:`
- Be careful to avoid infinite loops!

**Key Loop Concepts:**

**range() Function:**
- `range(start, stop, step)` generates a sequence of numbers
- `range(5)` generates 0, 1, 2, 3, 4
- `range(1, 6)` generates 1, 2, 3, 4, 5
- `range(0, 10, 2)` generates 0, 2, 4, 6, 8

**Loop Control Statements:**
- `break`: Immediately exits the loop
- `continue`: Skips the rest of the current iteration and continues with the next
- `else`: Executes when the loop completes normally (not via break)

**enumerate() Function:**
- Provides both index and value when iterating
- Useful when you need the position of items in a sequence

**Best Practices:**
- Use `for` loops for known iterations, `while` loops for conditional repetition
- Always ensure `while` loops have a way to terminate
- Use meaningful variable names in loops
- Consider using `enumerate()` when you need both index and value

The code below demonstrates various loop patterns and control mechanisms with practical examples.


````python
# for loop with range

while i in range(1,6):
    print(i)
    i = i + 2
````

````python
# for loop
fruits = ["apple", "banana", "orange", "grape"]
print("Fruits:")
for fruit in fruits:
    print(f"- {fruit}")

# for loop with range
print("\nNumbers 1 to 5:")
for i in range(1, 6):
    print(f"Number: {i}")

# while loop
count = 0
print("\nWhile loop countdown:")
while count < 5:
    print(f"Count: {count}")
    count += 1

# break and continue
print("\nBreak and continue example:")
for num in range(1, 11):
    if num == 3:
        continue  # Skip 3
    if num == 8:
        break     # Stop at 8
    print(f"Number: {num}")

# enumerate
print("\nEnumerate example:")
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
````

## 7. Functions: Reusable Code Blocks

Functions are one of the most important concepts in programming. They allow you to write code once and use it multiple times, making your programs more organized, readable, and maintainable.

**What are Functions?**
Functions are named blocks of code that perform a specific task. They can accept inputs (parameters) and return outputs.

**Key Function Concepts:**

**1. Function Definition:**
- Use `def` keyword followed by function name and parameters
- Structure: `def function_name(parameters):`

**2. Parameters and Arguments:**
- **Parameters**: Variables in the function definition
- **Arguments**: Values passed to the function when called
- Functions can have multiple parameters

**3. Default Parameters:**
- Provide default values for parameters
- Parameters with defaults must come after those without defaults

**4. Keyword Arguments:**
- Call functions using parameter names
- Allows flexibility in argument order
- Mix positional and keyword arguments (positional first)

**5. Return Values:**
- Use `return` statement to send data back
- Functions can return multiple values (as tuples)
- Functions without return statement return `None`

**6. Lambda Functions:**
- Anonymous (unnamed) functions
- Syntax: `lambda parameters: expression`
- Useful for simple, one-line functions

**Function Benefits:**
- **Reusability**: Write once, use many times
- **Modularity**: Break complex problems into smaller parts
- **Readability**: Give meaningful names to code blocks
- **Testing**: Test individual functions separately
- **Maintenance**: Update logic in one place

**Best Practices:**
- Use descriptive function names
- Keep functions focused on a single task
- Use docstrings to document function purpose
- Limit the number of parameters (3-4 is ideal)

The code below demonstrates various function types and patterns you'll use throughout your Python programming journey.


````python
# Basic function
def abc(name, age):
    age = age / 2
    return f"Hello, {name}, is {age} years old!"

abc("Jack", "abc")
````

````python
# Basic function
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))

# Function with multiple parameters
def calculate_area(length, width):
    return length * width

area = calculate_area(5, 3)
print(f"Area: {area}")

# Function with default parameters
def greet_with_title(name, title="Mr."):
    return f"Hello, {title} {name}!"

print(greet_with_title("Smith"))
print(greet_with_title("Johnson", "Dr."))

# Function with keyword arguments
def create_profile(name, age, city="Unknown", country="Unknown"):
    return f"Name: {name}, Age: {age}, City: {city}, Country: {country}"

profile = create_profile("Bob", 30, country="USA")
print(profile)

# Function returning multiple values
def get_name_parts(full_name):
    parts = full_name.split()
    return parts[0], parts[-1] if len(parts) > 1 else ""

first, last = get_name_parts("John Doe")
print(f"First: {first}, Last: {last}")

# Lambda function
square = lambda x: x ** 2
print(f"Square of 5: {square(5)}")

# Using lambda with built-in functions
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x ** 2, numbers))
print(f"Squared numbers: {squared_numbers}")
````

## 8. Importing Modules: Extending Python's Capabilities

Python's power comes from its extensive library of modules and packages. A module is a file containing Python code that can be imported and used in other programs. This allows you to leverage pre-written code for common tasks.

**What are Modules?**
- Modules are files containing Python definitions and statements
- They provide reusable code for specific functionality
- Python comes with many built-in modules (standard library)
- You can also install additional modules using pip

**Import Methods:**

**1. Import Entire Module:**
```python
import module_name
# Use: module_name.function_name()
```

**2. Import Specific Functions:**
```python
from module_name import function1, function2
# Use: function1() directly
```

**3. Import with Alias:**
```python
import module_name as alias
# Use: alias.function_name()
```

**4. Import All (Not Recommended):**
```python
from module_name import *
# Use: function_name() directly
```

**Common Built-in Modules:**
- `math`: Mathematical functions (sqrt, pi, sin, cos, etc.)
- `datetime`: Date and time operations
- `random`: Random number generation
- `os`: Operating system interface
- `sys`: System-specific parameters and functions

**Best Practices:**
- Import modules at the top of your file
- Use specific imports when you only need a few functions
- Use aliases for long module names
- Avoid `import *` as it can cause namespace pollution
- Use `dir(module)` to explore available functions in a module

**Why Modules Matter:**
- **Code Reusability**: Don't reinvent the wheel
- **Organization**: Keep related code together
- **Maintenance**: Update modules independently
- **Performance**: Optimized, tested code
- **Collaboration**: Share code across projects

The code below demonstrates different import methods and shows how to use some of Python's most useful built-in modules.


````python
from datetime import datetime, date
today = date.today()
print(f"{today}")
````

````python

# Import entire module
import math
print(f"Square root of 16: {math.sqrt(16)}")
print(f"Value of pi: {math.pi}")

# Import specific functions
from datetime import datetime, date
now = datetime.now()
today = date.today()
print(f"Current time: {now}")
print(f"Today's date: {today}")

# Import with alias
import random as rnd
random_number = rnd.randint(1, 10)
print(f"Random number: {random_number}")

# Import all from module (not recommended for large modules)
from math import *
print(f"Factorial of 5: {factorial(5)}")

# Check available functions in a module
print(f"\nMath module functions: {dir(math)[:10]}...")  # First 10 functions
````

## 9. Error Handling: Making Robust Programs

No program is perfect, and errors are inevitable. Python provides a powerful error handling mechanism that allows you to gracefully handle unexpected situations instead of letting your program crash.

**What are Exceptions?**
- Exceptions are errors that occur during program execution
- Python has many built-in exception types for different error conditions
- When an exception occurs, Python stops normal execution and looks for error handling code

**Common Exception Types:**
- `ZeroDivisionError`: Division by zero
- `TypeError`: Wrong data type used in operation
- `IndexError`: List index out of range
- `KeyError`: Dictionary key not found
- `FileNotFoundError`: File doesn't exist
- `ValueError`: Invalid value for operation

**Error Handling Structure:**

**1. try-except Block:**
```python
try:
    # Code that might cause an error
except ExceptionType:
    # Code to handle the error
```

**2. Multiple Exception Types:**
```python
try:
    # Code that might cause errors
except SpecificError:
    # Handle specific error
except AnotherError:
    # Handle another error
```

**3. try-except-else-finally:**
```python
try:
    # Code that might cause an error
except ExceptionType:
    # Handle the error
else:
    # Code that runs if no error occurred
finally:
    # Code that always runs (cleanup code)
```

**Error Handling Best Practices:**
- **Be Specific**: Catch specific exceptions, not generic ones
- **Fail Gracefully**: Provide meaningful error messages
- **Log Errors**: Record errors for debugging
- **Clean Up**: Use `finally` for cleanup operations
- **Don't Hide Errors**: Only catch exceptions you can handle
- **Test Error Conditions**: Verify your error handling works

**Why Error Handling Matters:**
- **User Experience**: Prevent crashes and provide helpful messages
- **Debugging**: Make it easier to identify and fix problems
- **Robustness**: Programs continue running despite unexpected input
- **Professional Code**: Shows attention to edge cases and error conditions

**Common Patterns:**
- Validate input before processing
- Use default values when operations fail
- Provide fallback behavior for critical functions
- Log errors for later analysis

The code below demonstrates various error handling patterns and shows how to create robust, user-friendly programs that handle unexpected situations gracefully.


````python
# Basic try-except
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Cannot divide by zero!"
    except TypeError:
        return "Invalid input types!"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "2"))

# Multiple exception types
def get_list_element(lst, index):
    try:
        return lst[index]
    except IndexError:
        return "Index out of range!"
    except TypeError:
        return "Invalid index type!"

numbers = [1, 2, 3, 4, 5]
print(get_list_element(numbers, 2))
print(get_list_element(numbers, 10))
print(get_list_element(numbers, "2"))

# try-except-else-finally
def process_file(filename):
    try:
        file = open(filename, 'r')
        content = file.read()
    except FileNotFoundError:
        print(f"File {filename} not found!")
        return None
    except PermissionError:
        print(f"Permission denied to read {filename}")
        return None
    else:
        print("File read successfully!")
        return content
    finally:
        print("File processing completed")

# This will show the error handling
result = process_file("nonexistent.txt")
````
