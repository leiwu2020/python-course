## Exercise 1: Data Preprocessing and Exploration (20 points)

**Task:** Preprocess data for machine learning and perform exploratory data analysis.

**Requirements:**
- Load and explore datasets
- Handle missing values and outliers
- Perform feature scaling and encoding
- Split data into training and testing sets


```python
# TODO: Complete the following tasks

# Import pandas library for data manipulation and analysis
import pandas as pd
# Import numpy library for numerical computing
import numpy as np
# Import machine learning datasets from scikit-learn
from sklearn.datasets import load_iris, load_wine
# Import train_test_split for splitting data into training and testing sets
from sklearn.model_selection import train_test_split
# Import preprocessing tools for data scaling and encoding
from sklearn.preprocessing import StandardScaler, LabelEncoder
# Import LogisticRegression for classification
from sklearn.linear_model import LogisticRegression
# Import RandomForestClassifier for ensemble learning
from sklearn.ensemble import RandomForestClassifier
# Import evaluation metrics for model performance assessment
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
# Import matplotlib for plotting and visualization
import matplotlib.pyplot as plt
# Import seaborn for statistical data visualization
import seaborn as sns

# 1. Load and explore datasets
# Print a header for the dataset loading and exploration section
print("=== DATASET LOADING AND EXPLORATION ===")

# Load Iris dataset
# Load the famous Iris dataset from scikit-learn
iris = load_iris()
# Convert the iris data to a pandas DataFrame with feature names as columns
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
# Add the target variable (species labels as numbers) to the DataFrame
iris_df['target'] = iris.target
# Add the target variable (species labels as names) to the DataFrame
iris_df['species'] = iris.target_names[iris.target]

# Print the shape (dimensions) of the iris dataset
print(f"Iris dataset shape: {iris_df.shape}")
# Print the first 5 rows of the iris dataset
print(f"\nFirst 5 rows:\n{iris_df.head()}")
# Print dataset information including data types and memory usage
print(f"\nDataset info:")
print(iris_df.info())
# Print statistical summary of the numerical columns
print(f"\nStatistical summary:\n{iris_df.describe()}")

# TODO: Load Wine dataset
# Load the Wine dataset from scikit-learn
wine = load_wine()
# Convert the wine data to a pandas DataFrame with feature names as columns
wine_df = pd.DataFrame(wine.data, columns=wine.feature_names)
# Add the target variable (wine class labels as numbers) to the DataFrame
wine_df['target'] = wine.target
# Add the target variable (wine class labels as names) to the DataFrame
wine_df['wine_type'] = wine.target_names[wine.target]

# Print the shape (dimensions) of the wine dataset
print(f"\nWine dataset shape: {wine_df.shape}")
# Print the first 5 rows of the wine dataset
print(f"\nFirst 5 rows:\n{wine_df.head()}")

# 2. Handle missing values and outliers
# Print a header for the data cleaning section
print(f"\n=== DATA CLEANING ===")

# Check for missing values
# Count missing values in each column of the iris dataset
print(f"Missing values in Iris dataset:\n{iris_df.isnull().sum()}")
# Count missing values in each column of the wine dataset
print(f"Missing values in Wine dataset:\n{wine_df.isnull().sum()}")

# TODO: Handle outliers using IQR method
# Define a function to detect outliers using the Interquartile Range (IQR) method
def detect_outliers_iqr(data, column):
    # Calculate the first quartile (25th percentile)
    Q1 = data[column].quantile(0.25)
    # Calculate the third quartile (75th percentile)
    Q3 = data[column].quantile(0.75)
    # Calculate the Interquartile Range (IQR)
    IQR = Q3 - Q1
    # Calculate the lower bound for outliers (Q1 - 1.5 * IQR)
    lower_bound = Q1 - 1.5 * IQR
    # Calculate the upper bound for outliers (Q3 + 1.5 * IQR)
    upper_bound = Q3 + 1.5 * IQR
    # Find rows where the column value is outside the bounds (outliers)
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    # Return the outliers
    return outliers

# Check for outliers in numerical columns
# Select only numerical columns from the iris dataset
numerical_columns = iris_df.select_dtypes(include=[np.number]).columns
# Print a header for outlier detection
print(f"\nOutlier detection for Iris dataset:")
# Iterate through each numerical column (excluding target)
for col in numerical_columns:
    # Skip the target column as it's not a feature
    if col != 'target':
        # Detect outliers in the current column
        outliers = detect_outliers_iqr(iris_df, col)
        # Print the number of outliers found in this column
        print(f"{col}: {len(outliers)} outliers")

# 3. Perform feature scaling and encoding
# Print a header for the feature scaling and encoding section
print(f"\n=== FEATURE SCALING AND ENCODING ===")

# Separate features and target
# Drop the target and species columns to get only the feature columns
X_iris = iris_df.drop(['target', 'species'], axis=1)
# Extract the target variable (species labels as numbers)
y_iris = iris_df['target']

# TODO: Apply StandardScaler
# Create an instance of StandardScaler for feature normalization
scaler = StandardScaler()
# Fit the scaler to the training data and transform the features
X_iris_scaled = scaler.fit_transform(X_iris)
# Convert the scaled features back to a DataFrame with original column names
X_iris_scaled_df = pd.DataFrame(X_iris_scaled, columns=X_iris.columns)

# Print the original features (first 5 rows)
print(f"Original features (first 5 rows):\n{X_iris.head()}")
# Print the scaled features (first 5 rows)
print(f"Scaled features (first 5 rows):\n{X_iris_scaled_df.head()}")

# 4. Split data into training and testing sets
# Print a header for the data splitting section
print(f"\n=== DATA SPLITTING ===")

# TODO: Split the data
# Split the data into training and testing sets with 80/20 split
X_train, X_test, y_train, y_test = train_test_split(
    X_iris_scaled, y_iris, test_size=0.2, random_state=42, stratify=y_iris
)

# Print the shape of the training set
print(f"Training set shape: {X_train.shape}")
# Print the shape of the test set
print(f"Test set shape: {X_test.shape}")
# Print the distribution of target classes in training set
print(f"Training target distribution: {np.bincount(y_train)}")
# Print the distribution of target classes in test set
print(f"Test target distribution: {np.bincount(y_test)}")

# 5. Train and evaluate models
# Print a header for the model training and evaluation section
print(f"\n=== MODEL TRAINING AND EVALUATION ===")

# Logistic Regression
# Create an instance of LogisticRegression with random state for reproducibility
lr_model = LogisticRegression(random_state=42)
# Train the logistic regression model on the training data
lr_model.fit(X_train, y_train)
# Make predictions on the test set using the trained model
lr_pred = lr_model.predict(X_test)
# Calculate the accuracy of the logistic regression model
lr_accuracy = accuracy_score(y_test, lr_pred)

# Print the accuracy of the logistic regression model
print(f"Logistic Regression Accuracy: {lr_accuracy:.4f}")

# TODO: Random Forest
# Create an instance of RandomForestClassifier with 100 trees and random state for reproducibility
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
# Train the random forest model on the training data
rf_model.fit(X_train, y_train)
# Make predictions on the test set using the trained random forest model
rf_pred = rf_model.predict(X_test)
# Calculate the accuracy of the random forest model
rf_accuracy = accuracy_score(y_test, rf_pred)

# Print the accuracy of the random forest model
print(f"Random Forest Accuracy: {rf_accuracy:.4f}")

# Model comparison
# Print a header for model comparison
print(f"\nModel Comparison:")
# Print the accuracy of logistic regression
print(f"Logistic Regression: {lr_accuracy:.4f}")
# Print the accuracy of random forest
print(f"Random Forest: {rf_accuracy:.4f}")

# Classification report
# Print a header for the classification report
print(f"\nRandom Forest Classification Report:")
# Print detailed classification metrics including precision, recall, and F1-score
print(classification_report(y_test, rf_pred, target_names=iris.target_names))
```
